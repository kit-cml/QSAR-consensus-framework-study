```python
import os
import ast
import numpy as np
import pandas as pd
from rdkit import DataStructs
import matplotlib.pyplot as plt

# ======================================
# Konfigurasi umum
# ======================================
base_output_dir = r"C:\\Fauzan\\Manuskrip QSAR 3\\Toxicology Mechanism and Method\\Revisi 1\\AD revisi"
os.makedirs(base_output_dir, exist_ok=True)

# Daftar endpoint & datasets (sama seperti di kode kamu)
endpoints = [
    {
        "endpoint": "Drug Induced Nephrotoxicity",
        "file_tag": "Drug_Induced_Nephrotoxicity",
        "datasets": [
            {
                "label": "Drug Induced Nephrotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Drug Induced Nephrotoxicity\Train_Set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Drug Induced Nephrotoxicity – External Validation",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Drug Induced Nephrotoxicity\External Validation_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Drug Induced Nephrotoxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Drug Induced Nephrotoxicity\Test_Set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "AMES Mutagenicity",
        "file_tag": "AMES_Mutagenicity",
        "datasets": [
            {
                "label": "AMES Mutagenicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\AMES\AMES_Train set.xlsx"
            },
            {
                "label": "AMES Mutagenicity – External Validation 1",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\AMES\AMES_Set1_external_validation.xlsx"
            },
            {
                "label": "AMES Mutagenicity – External Validation 2",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\AMES\AMES_Set2_external_validation.xlsx"
            },
        ]
    },
    {
        "endpoint": "Developmental Toxicity",
        "file_tag": "Developmental_Toxicity",
        "datasets": [
            {
                "label": "Developmental Toxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Development Toxicity\Dev_Train set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Developmental Toxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Development Toxicity\Dev_Test_set_balanced_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Hepatotoxicity",
        "file_tag": "Hepatotoxicity",
        "datasets": [
            {
                "label": "Hepatotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\HEP_Train_set_train_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Hepatotoxicity – External Validation 1",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\Test_set_ValRandom_fingerprints_with_RDKit_CDK.xlsx"
            },
            {
                "label": "Hepatotoxicity – External Validation 2",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\Ext_Val_ValBLACK_blackNE_.xlsx"
            },
        ]
    },
    {
        "endpoint": "Neurotoxicity",
        "file_tag": "Neurotoxicity",
        "datasets": [
            {
                "label": "Neurotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\NeuroToxicity\Train_Set_with_fingerprints_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Neurotoxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\NeuroToxicity\Validation_Set_with_fingerprints_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Respiratory Toxicity",
        "file_tag": "Respiratory_Toxicity",
        "datasets": [
            {
                "label": "Respiratory Toxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Respiratory Toxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_Test_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Respiratory Toxicity – External Validation",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_External_validation_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Skin Sensitization",
        "file_tag": "Skin_Sensitization",
        "datasets": [
            {
                "label": "Skin Sensitization – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Skin Sensinization\SkinSen_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Skin Sensitization – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Skin Sensinization\SkinSen_Test_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Cardiotoxicity",
        "file_tag": "Cardiotoxicity",
        "datasets": [
            {
                "label": "Cardiotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 1",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 1_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 2",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 2_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 3",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 3_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
]

# Grid cutoff: 0.30, 0.35, ..., 0.70
cutoff_grid = [round(x, 2) for x in np.arange(0.30, 0.71, 0.05)]

# Coverage minimum untuk memilih cutoff (mis. 90%)
coverage_min = 75.0

fp_cols = ["Morgan_Descriptors", "MACCS_Descriptors", "APF_Descriptors"]

# ======================================
# Fungsi bantu
# ======================================
def convert_to_bitvect(fp_list, n_bits=None):
    if isinstance(fp_list, str):
        try:
            fp_list = ast.literal_eval(fp_list)
        except Exception:
            fp_list = []
    if fp_list is None or len(fp_list) == 0:
        if n_bits is None:
            raise ValueError("n_bits harus ditentukan jika fingerprint kosong")
        fp_list = [0] * n_bits
    if n_bits is None:
        n_bits = len(fp_list)
    bv = DataStructs.ExplicitBitVect(n_bits)
    for i, bit in enumerate(fp_list):
        if bit:
            bv.SetBit(i)
    return bv

def max_similarity_to_train_fast(fp_test_list, fp_train_list):
    train_bvs = [convert_to_bitvect(fp) for fp in fp_train_list]
    if len(train_bvs) == 0:
        return np.zeros(len(fp_test_list))
    n_bits = len(train_bvs[0])
    max_sims = []
    for fp_test in fp_test_list:
        test_bv = convert_to_bitvect(fp_test, n_bits=n_bits)
        sims = DataStructs.BulkTanimotoSimilarity(test_bv, train_bvs)
        max_sims.append(max(sims) if len(sims) > 0 else 0.0)
    return np.array(max_sims)

def leverage_physchem(X_train, X_test):
    X_train = np.array(X_train, dtype=float)
    X_test = np.array(X_test, dtype=float)
    if X_train.shape[1] == 0:
        n_test = X_test.shape[0]
        return np.zeros(n_test), np.nan, np.ones(n_test, dtype=bool)
    n_samples, n_features = X_train.shape
    X_train_aug = np.hstack([np.ones((n_samples, 1)), X_train])
    X_test_aug = np.hstack([np.ones((X_test.shape[0], 1)), X_test])
    XtX = X_train_aug.T @ X_train_aug
    XtX_inv = np.linalg.pinv(XtX)
    H_test = np.sum((X_test_aug @ XtX_inv) * X_test_aug, axis=1)
    h_star = 3 * (n_features + 1) / n_samples
    return H_test, h_star, H_test <= h_star

# ======================================
# 1) Loop per endpoint
# ======================================
for ep in endpoints:
    endpoint_name = ep["endpoint"]
    file_tag = ep["file_tag"]
    datasets = ep["datasets"]

    output_dir = os.path.join(base_output_dir, file_tag)
    os.makedirs(output_dir, exist_ok=True)

    print("\n==============================")
    print(f"Endpoint: {endpoint_name}")
    print("==============================")

    # 1.1 TRAIN set
    train_file = datasets[0]["file"]
    print(f"Loading TRAIN set: {train_file}")
    train_df = pd.read_excel(train_file)

    exclude_cols = ["SMILES", "Outcome", "RDK_Descriptors"] + fp_cols
    physchem_cols = [
        col for col in train_df.columns
        if (col not in exclude_cols) and np.issubdtype(train_df[col].dtype, np.number)
    ]
    print("Physchem cols used for leverage:")
    print(physchem_cols)

    train_fps_dict = {fp_name: train_df[fp_name].tolist() for fp_name in fp_cols}

    ep_coverage_records = []
    per_dataset_results = {}  # simpan df + max_sim_dict + physchem_in_ad per dataset

    # 1.2 Semua datasets (train + eval)
    for ds in datasets:
        ds_label = ds["label"]
        file_path = ds["file"]
        print(f"\nProcessing dataset: {ds_label}")
        print(f"File: {file_path}")

        df = pd.read_excel(file_path)
        base_name = os.path.splitext(os.path.basename(file_path))[0]

        # Max Tanimoto
        max_sim_dict = {}
        for fp_name in fp_cols:
            print(f"  > Computing max Tanimoto for {fp_name} ...")
            test_fps = df[fp_name].tolist()
            max_sims = max_similarity_to_train_fast(test_fps, train_fps_dict[fp_name])
            max_sim_dict[fp_name] = max_sims
            df[f"{fp_name}_MaxSimTrain"] = max_sims

        # Leverage physchem
        print("  > Computing leverage for physicochemical descriptors ...")
        train_phys = train_df[physchem_cols].apply(pd.to_numeric, errors="coerce")
        test_phys = df[physchem_cols].apply(pd.to_numeric, errors="coerce")
        train_means = train_phys.mean()
        train_phys = train_phys.fillna(train_means)
        test_phys = test_phys.fillna(train_means)

        X_train = train_phys.values
        X_test = test_phys.values
        H_test, h_star, physchem_in_ad = leverage_physchem(X_train, X_test)
        df["Physchem_Leverage"] = H_test
        df["Physchem_IN_AD"] = physchem_in_ad

        # Simpan untuk dipakai lagi saat apply cutoff akhir
        per_dataset_results[ds_label] = {
            "df": df,
            "base_name": base_name,
            "max_sim_dict": max_sim_dict,
            "physchem_in_ad": physchem_in_ad,
        }

        # Sensitivity analysis coverage
        for t in cutoff_grid:
            record = {
                "Endpoint": endpoint_name,
                "File_tag": file_tag,
                "Dataset_label": ds_label,
                "Base_name": base_name,
                "Cutoff": t,
                "Total_compounds": len(df),
            }
            for fp_name in fp_cols:
                in_ad_mask = max_sim_dict[fp_name] >= t
                record[f"{fp_name}_IN_AD_pct"] = in_ad_mask.mean() * 100.0
            record["Physchem_IN_AD_pct"] = df["Physchem_IN_AD"].mean() * 100.0
            consensus_in_ad = (
                (max_sim_dict["Morgan_Descriptors"] >= t).astype(int)
                + (max_sim_dict["MACCS_Descriptors"] >= t).astype(int)
                + (max_sim_dict["APF_Descriptors"] >= t).astype(int)
                + df["Physchem_IN_AD"].astype(int)
            ) >= 3
            record["Consensus_IN_AD_pct"] = consensus_in_ad.mean() * 100.0
            ep_coverage_records.append(record)

    # 1.3 Summary coverage per endpoint
    ep_cov_df = pd.DataFrame(ep_coverage_records)
    ep_summary_path = os.path.join(
        output_dir,
        f"AD_sensitivity_coverage_summary_{file_tag}.xlsx"
    )
    ep_cov_df.to_excel(ep_summary_path, index=False)
    print(f"\n✅ Saved AD sensitivity coverage summary for {endpoint_name}: {ep_summary_path}")

    # 1.4 Pilih tau_endpoint berbasis consensus coverage (eval only)
    eval_rows = ~ep_cov_df["Dataset_label"].str.contains("Train Set", case=False)
    eval_cov = ep_cov_df[eval_rows]
    # rata-rata consensus coverage per cutoff di seluruh test/external set endpoint ini
    avg_cov_by_t = (
        eval_cov.groupby("Cutoff")["Consensus_IN_AD_pct"]
        .mean()
        .reset_index()
        .sort_values("Cutoff")
    )

    # ambil cutoff tertinggi yang masih >= coverage_min
    valid = avg_cov_by_t[avg_cov_by_t["Consensus_IN_AD_pct"] >= coverage_min]
    if len(valid) == 0:
        tau_endpoint = cutoff_grid[0]
        print(f"⚠ No cutoff reached {coverage_min}% coverage; using lowest cutoff {tau_endpoint:.2f}")
    else:
        tau_endpoint = valid["Cutoff"].max()
        print(f"Selected tau_endpoint for {endpoint_name}: {tau_endpoint:.2f}")

    # Simpan info threshold per endpoint
    with open(os.path.join(output_dir, f"{file_tag}_selected_tau.txt"), "w") as f:
        f.write(f"Selected Tanimoto cutoff (tau_endpoint) for {endpoint_name}: {tau_endpoint:.2f}\n")
        f.write(f"Minimum consensus coverage required: {coverage_min:.1f}%\n")

    # 1.5 Apply tau_endpoint untuk flag IN_AD per senyawa dan simpan file
    for ds_label, info in per_dataset_results.items():
        df = info["df"]
        base_name = info["base_name"]
        max_sim_dict = info["max_sim_dict"]

        print(f"\nApplying tau_endpoint = {tau_endpoint:.2f} to dataset: {ds_label}")

        for fp_name in fp_cols:
            df[f"{fp_name}_IN_AD"] = max_sim_dict[fp_name] >= tau_endpoint

        df["Consensus_IN_AD"] = (
            df["Morgan_Descriptors_IN_AD"].astype(int)
            + df["MACCS_Descriptors_IN_AD"].astype(int)
            + df["APF_Descriptors_IN_AD"].astype(int)
            + df["Physchem_IN_AD"].astype(int)
        ) >= 3

        # Log coverage pada tau_endpoint
        morgan_cov = df["Morgan_Descriptors_IN_AD"].mean() * 100.0
        maccs_cov = df["MACCS_Descriptors_IN_AD"].mean() * 100.0
        apf_cov = df["APF_Descriptors_IN_AD"].mean() * 100.0
        physchem_cov = df["Physchem_IN_AD"].mean() * 100.0
        consensus_cov = df["Consensus_IN_AD"].mean() * 100.0

        print(f"  > Coverage at tau_endpoint = {tau_endpoint:.2f}:")
        print(f"    - Morgan IN_AD      : {morgan_cov:.1f}%")
        print(f"    - MACCS IN_AD       : {maccs_cov:.1f}%")
        print(f"    - APF IN_AD         : {apf_cov:.1f}%")
        print(f"    - Physchem IN_AD    : {physchem_cov:.1f}%")
        print(f"    - Consensus IN_AD   : {consensus_cov:.1f}%")

        output_path = os.path.join(
            output_dir,
            f"{file_tag}_{base_name}_AD_results_with_tau_{tau_endpoint:.2f}.xlsx"
        )
        df.to_excel(output_path, index=False)
        print(f"  > Saved per-compound AD results: {output_path}")

    # 1.6 Plot coverage vs cutoff (Consensus) per dataset (optional, masih pakai full ep_cov_df)
    for ds_label in ep_cov_df["Dataset_label"].unique():
        sub = ep_cov_df[ep_cov_df["Dataset_label"] == ds_label].sort_values("Cutoff")
        plt.figure(figsize=(6, 4))
        plt.plot(sub["Cutoff"], sub["Consensus_IN_AD_pct"],
                 marker="o", linestyle="-", color="tab:blue",
                 label="Consensus IN_AD %")
        plt.axvline(tau_endpoint, color="red", linestyle="--",
                    label=f"Selected tau = {tau_endpoint:.2f}")
        plt.ylim(0, 100)
        plt.xlabel("Tanimoto cutoff (t)")
        plt.ylabel("Consensus AD coverage (%)")
        plt.title(f"{endpoint_name}\n{ds_label}")
        plt.grid(True, alpha=0.3)
        ax = plt.gca()
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        plt.legend(frameon=False)
        plt.tight_layout()
        safe_ds = ds_label.replace(" ", "_").replace("–", "-").replace(":", "").replace("/", "_")
        plot_path = os.path.join(
            output_dir,
            f"{file_tag}_{safe_ds}_Consensus_AD_coverage_vs_cutoff_with_tau.png"
        )
        plt.savefig(plot_path, dpi=300)
        plt.close()
        print(f"  > Saved plot: {plot_path}")

print("\n✅ All endpoints AD processing finished (with data-driven tau_endpoint per endpoint).")
```

```python
import os
import ast
import numpy as np
import pandas as pd
from rdkit import DataStructs
import matplotlib.pyplot as plt

# ======================================
# Konfigurasi umum
# ======================================
base_output_dir = r"C:\\Fauzan\\Manuskrip QSAR 3\\Toxicology Mechanism and Method\\Revisi 1\\AD revisi_rule2"
os.makedirs(base_output_dir, exist_ok=True)

# Daftar endpoint & datasets (sama seperti di kode kamu)
endpoints = [
    {
        "endpoint": "Drug Induced Nephrotoxicity",
        "file_tag": "Drug_Induced_Nephrotoxicity",
        "datasets": [
            {
                "label": "Drug Induced Nephrotoxicity – Train Set",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\Drug Induced Nephrotoxicity\\Train_Set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Drug Induced Nephrotoxicity – External Validation",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\Drug Induced Nephrotoxicity\\External Validation_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Drug Induced Nephrotoxicity – Unseen Test Set",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\Drug Induced Nephrotoxicity\\Test_Set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "AMES Mutagenicity",
        "file_tag": "AMES_Mutagenicity",
        "datasets": [
            {
                "label": "AMES Mutagenicity – Train Set",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\AMES\\AMES_Train set.xlsx"
            },
            {
                "label": "AMES Mutagenicity – External Validation 1",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\AMES\\AMES_Set1_external_validation.xlsx"
            },
            {
                "label": "AMES Mutagenicity – External Validation 2",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\AMES\\AMES_Set2_external_validation.xlsx"
            },
        ]
    },
    {
        "endpoint": "Developmental Toxicity",
        "file_tag": "Developmental_Toxicity",
        "datasets": [
            {
                "label": "Developmental Toxicity – Train Set",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\Development Toxicity\\Dev_Train set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Developmental Toxicity – Unseen Test Set",
                "file": r"C:\\Fauzan\\Manuscripts QSAR-RA 2\\Manuscript\\AD Analysis\\Development Toxicity\\Dev_Test_set_balanced_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
     {
        "endpoint": "Hepatotoxicity",
        "file_tag": "Hepatotoxicity",
        "datasets": [
            {
                "label": "Hepatotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\HEP_Train_set_train_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Hepatotoxicity – External Validation 1",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\Test_set_ValRandom_fingerprints_with_RDKit_CDK.xlsx"
            },
            {
                "label": "Hepatotoxicity – External Validation 2",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Hepatoxicity\Ext_Val_ValBLACK_blackNE_.xlsx"
            },
        ]
    },
    {
        "endpoint": "Neurotoxicity",
        "file_tag": "Neurotoxicity",
        "datasets": [
            {
                "label": "Neurotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\NeuroToxicity\Train_Set_with_fingerprints_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Neurotoxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\NeuroToxicity\Validation_Set_with_fingerprints_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Respiratory Toxicity",
        "file_tag": "Respiratory_Toxicity",
        "datasets": [
            {
                "label": "Respiratory Toxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Respiratory Toxicity – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_Test_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Respiratory Toxicity – External Validation",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Respiratory Tox\Resp_External_validation_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Skin Sensitization",
        "file_tag": "Skin_Sensitization",
        "datasets": [
            {
                "label": "Skin Sensitization – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Skin Sensinization\SkinSen_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Skin Sensitization – Unseen Test Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Skin Sensinization\SkinSen_Test_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
    {
        "endpoint": "Cardiotoxicity",
        "file_tag": "Cardiotoxicity",
        "datasets": [
            {
                "label": "Cardiotoxicity – Train Set",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Train_set_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 1",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 1_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 2",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 2_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
            {
                "label": "Cardiotoxicity – External Validation 3",
                "file": r"C:\Fauzan\Manuscripts QSAR-RA 2\Manuscript\AD Analysis\Cardiotoxicity\CardioTox_Test_set 3_with_fingerprints_sorted_with_RDKit_and_CDK_features.xlsx"
            },
        ]
    },
]

# Grid cutoff: 0.30, 0.35, ..., 0.70
cutoff_grid = [round(x, 2) for x in np.arange(0.30, 0.71, 0.05)]

# Coverage minimum untuk memilih cutoff (mis. 90%)
coverage_min = 75.0

fp_cols = ["Morgan_Descriptors", "MACCS_Descriptors", "APF_Descriptors"]

# ======================================
# Fungsi bantu
# ======================================
def convert_to_bitvect(fp_list, n_bits=None):
    if isinstance(fp_list, str):
        try:
            fp_list = ast.literal_eval(fp_list)
        except Exception:
            fp_list = []
    if fp_list is None or len(fp_list) == 0:
        if n_bits is None:
            raise ValueError("n_bits harus ditentukan jika fingerprint kosong")
        fp_list = [0] * n_bits
    if n_bits is None:
        n_bits = len(fp_list)
    bv = DataStructs.ExplicitBitVect(n_bits)
    for i, bit in enumerate(fp_list):
        if bit:
            bv.SetBit(i)
    return bv

def max_similarity_to_train_fast(fp_test_list, fp_train_list):
    train_bvs = [convert_to_bitvect(fp) for fp in fp_train_list]
    if len(train_bvs) == 0:
        return np.zeros(len(fp_test_list))
    n_bits = len(train_bvs[0])
    max_sims = []
    for fp_test in fp_test_list:
        test_bv = convert_to_bitvect(fp_test, n_bits=n_bits)
        sims = DataStructs.BulkTanimotoSimilarity(test_bv, train_bvs)
        max_sims.append(max(sims) if len(sims) > 0 else 0.0)
    return np.array(max_sims)

def leverage_physchem(X_train, X_test):
    X_train = np.array(X_train, dtype=float)
    X_test = np.array(X_test, dtype=float)
    if X_train.shape[1] == 0:
        n_test = X_test.shape[0]
        return np.zeros(n_test), np.nan, np.ones(n_test, dtype=bool)
    n_samples, n_features = X_train.shape
    X_train_aug = np.hstack([np.ones((n_samples, 1)), X_train])
    X_test_aug = np.hstack([np.ones((X_test.shape[0], 1)), X_test])
    XtX = X_train_aug.T @ X_train_aug
    XtX_inv = np.linalg.pinv(XtX)
    H_test = np.sum((X_test_aug @ XtX_inv) * X_test_aug, axis=1)
    h_star = 3 * (n_features + 1) / n_samples
    return H_test, h_star, H_test <= h_star

# ======================================
# 1) Loop per endpoint
# ======================================
for ep in endpoints:
    endpoint_name = ep["endpoint"]
    file_tag = ep["file_tag"]
    datasets = ep["datasets"]

    output_dir = os.path.join(base_output_dir, file_tag)
    os.makedirs(output_dir, exist_ok=True)

    print("\n==============================")
    print(f"Endpoint: {endpoint_name}")
    print("==============================")

    # 1.1 TRAIN set
    train_file = datasets[0]["file"]
    print(f"Loading TRAIN set: {train_file}")
    train_df = pd.read_excel(train_file)

    exclude_cols = ["SMILES", "Outcome", "RDK_Descriptors"] + fp_cols
    physchem_cols = [
        col for col in train_df.columns
        if (col not in exclude_cols) and np.issubdtype(train_df[col].dtype, np.number)
    ]
    print("Physchem cols used for leverage:")
    print(physchem_cols)

    train_fps_dict = {fp_name: train_df[fp_name].tolist() for fp_name in fp_cols}

    ep_coverage_records = []
    per_dataset_results = {}  # simpan df + max_sim_dict + physchem_in_ad per dataset

    # 1.2 Semua datasets (train + eval)
    for ds in datasets:
        ds_label = ds["label"]
        file_path = ds["file"]
        print(f"\nProcessing dataset: {ds_label}")
        print(f"File: {file_path}")

        df = pd.read_excel(file_path)
        base_name = os.path.splitext(os.path.basename(file_path))[0]

        # Max Tanimoto
        max_sim_dict = {}
        for fp_name in fp_cols:
            print(f"  > Computing max Tanimoto for {fp_name} ...")
            test_fps = df[fp_name].tolist()
            max_sims = max_similarity_to_train_fast(test_fps, train_fps_dict[fp_name])
            max_sim_dict[fp_name] = max_sims
            df[f"{fp_name}_MaxSimTrain"] = max_sims

        # Leverage physchem
        print("  > Computing leverage for physicochemical descriptors ...")
        train_phys = train_df[physchem_cols].apply(pd.to_numeric, errors="coerce")
        test_phys = df[physchem_cols].apply(pd.to_numeric, errors="coerce")
        train_means = train_phys.mean()
        train_phys = train_phys.fillna(train_means)
        test_phys = test_phys.fillna(train_means)

        X_train = train_phys.values
        X_test = test_phys.values
        H_test, h_star, physchem_in_ad = leverage_physchem(X_train, X_test)
        df["Physchem_Leverage"] = H_test
        df["Physchem_IN_AD"] = physchem_in_ad

        # Simpan untuk dipakai lagi saat apply cutoff akhir
        per_dataset_results[ds_label] = {
            "df": df,
            "base_name": base_name,
            "max_sim_dict": max_sim_dict,
            "physchem_in_ad": physchem_in_ad,
        }

        # Sensitivity analysis coverage
        for t in cutoff_grid:
            record = {
                "Endpoint": endpoint_name,
                "File_tag": file_tag,
                "Dataset_label": ds_label,
                "Base_name": base_name,
                "Cutoff": t,
                "Total_compounds": len(df),
            }
            for fp_name in fp_cols:
                in_ad_mask = max_sim_dict[fp_name] >= t
                record[f"{fp_name}_IN_AD_pct"] = in_ad_mask.mean() * 100.0
            record["Physchem_IN_AD_pct"] = df["Physchem_IN_AD"].mean() * 100.0
            consensus_in_ad = (
                (max_sim_dict["Morgan_Descriptors"] >= t).astype(int)
                + (max_sim_dict["MACCS_Descriptors"] >= t).astype(int)
                + (max_sim_dict["APF_Descriptors"] >= t).astype(int)
                + df["Physchem_IN_AD"].astype(int)
            ) >= 3
            record["Consensus_IN_AD_pct"] = consensus_in_ad.mean() * 100.0
            ep_coverage_records.append(record)

    # 1.3 Summary coverage per endpoint
    ep_cov_df = pd.DataFrame(ep_coverage_records)
    ep_summary_path = os.path.join(
        output_dir,
        f"AD_sensitivity_coverage_summary_{file_tag}.xlsx"
    )
    ep_cov_df.to_excel(ep_summary_path, index=False)
    print(f"\n✅ Saved AD sensitivity coverage summary for {endpoint_name}: {ep_summary_path}")

    # 1.4 Pilih tau_endpoint berbasis consensus coverage (eval only)
    eval_rows = ~ep_cov_df["Dataset_label"].str.contains("Train Set", case=False)
    eval_cov = ep_cov_df[eval_rows]
    # rata-rata consensus coverage per cutoff di seluruh test/external set endpoint ini
    avg_cov_by_t = (
        eval_cov.groupby("Cutoff")["Consensus_IN_AD_pct"]
        .mean()
        .reset_index()
        .sort_values("Cutoff")
    )

    # ambil cutoff tertinggi yang masih >= coverage_min
    valid = avg_cov_by_t[avg_cov_by_t["Consensus_IN_AD_pct"] >= coverage_min]
    if len(valid) == 0:
        tau_endpoint_raw = cutoff_grid[0]
        print(f"⚠ No cutoff reached {coverage_min}% coverage; using lowest cutoff {tau_endpoint_raw:.2f}")
    else:
        tau_endpoint_raw = valid["Cutoff"].max()
        print(f"Selected tau_endpoint for {endpoint_name}: {tau_endpoint_raw:.2f}")

    # Terapkan aturan minimum cutoff 0.50
    tau_min = 0.50
    tau_endpoint = max(tau_endpoint_raw, tau_min)
    if tau_endpoint > tau_endpoint_raw:
        print(f"Applied minimum cutoff rule: tau_endpoint raised from {tau_endpoint_raw:.2f} to {tau_endpoint:.2f}")
    else:
        print(f"tau_endpoint kept at {tau_endpoint:.2f} (≥ {tau_min:.2f})")

    # Simpan info threshold per endpoint (UTF-8)
    tau_log_path = os.path.join(output_dir, f"{file_tag}_selected_tau.txt")
    with open(tau_log_path, "w", encoding="utf-8") as f:
        f.write(f"Selected Tanimoto cutoff (tau_endpoint) for {endpoint_name}: {tau_endpoint:.2f}\n")
        f.write(f"Minimum consensus coverage required: {coverage_min:.1f}%\n")
        f.write(f"Applied minimum cutoff rule (tau_min = {tau_min:.2f})\n\n")

    # 1.5 Apply tau_endpoint untuk flag IN_AD per senyawa dan simpan file
    eval_summary_lines = []  # ringkasan untuk evaluation set

    for ds_label, info in per_dataset_results.items():
        df = info["df"]
        base_name = info["base_name"]
        max_sim_dict = info["max_sim_dict"]

        print(f"\nApplying tau_endpoint = {tau_endpoint:.2f} to dataset: {ds_label}")

        for fp_name in fp_cols:
            df[f"{fp_name}_IN_AD"] = max_sim_dict[fp_name] >= tau_endpoint

        df["Consensus_IN_AD"] = (
            df["Morgan_Descriptors_IN_AD"].astype(int)
            + df["MACCS_Descriptors_IN_AD"].astype(int)
            + df["APF_Descriptors_IN_AD"].astype(int)
            + df["Physchem_IN_AD"].astype(int)
        ) >= 2

        # Log coverage pada tau_endpoint
        morgan_cov = df["Morgan_Descriptors_IN_AD"].mean() * 100.0
        maccs_cov = df["MACCS_Descriptors_IN_AD"].mean() * 100.0
        apf_cov = df["APF_Descriptors_IN_AD"].mean() * 100.0
        physchem_cov = df["Physchem_IN_AD"].mean() * 100.0
        consensus_cov = df["Consensus_IN_AD"].mean() * 100.0

        print(f"  > Coverage at tau_endpoint = {tau_endpoint:.2f}:")
        print(f"    - Morgan IN_AD      : {morgan_cov:.1f}%")
        print(f"    - MACCS IN_AD       : {maccs_cov:.1f}%")
        print(f"    - APF IN_AD         : {apf_cov:.1f}%")
        print(f"    - Physchem IN_AD    : {physchem_cov:.1f}%")
        print(f"    - Consensus IN_AD   : {consensus_cov:.1f}%")

        # Kalau evaluation set (bukan Train Set), simpan ringkasannya
        if "Train Set" not in ds_label:
            safe_label = ds_label.replace("–", "-")
            eval_summary_lines.append(
                f"[{safe_label}]\n"
                f"  N compounds          : {len(df)}\n"
                f"  Morgan IN_AD (%)     : {morgan_cov:.1f}\n"
                f"  MACCS IN_AD (%)      : {maccs_cov:.1f}\n"
                f"  APF IN_AD (%)        : {apf_cov:.1f}\n"
                f"  Physchem IN_AD (%)   : {physchem_cov:.1f}\n"
                f"  Consensus IN_AD (%)  : {consensus_cov:.1f}\n\n"
            )

        output_path = os.path.join(
            output_dir,
            f"{file_tag}_{base_name}_AD_results_with_tau_{tau_endpoint:.2f}.xlsx"
        )
        df.to_excel(output_path, index=False)
        print(f"  > Saved per-compound AD results: {output_path}")

    # 1.5b Tulis ringkasan evaluation set ke log txt yang sama (UTF-8)
    if eval_summary_lines:
        with open(tau_log_path, "a", encoding="utf-8") as f:
            f.write("Evaluation-set coverage summary at selected tau_endpoint:\n")
            f.write(f"(Only non-train datasets are listed; tau_endpoint = {tau_endpoint:.2f})\n\n")
            for line in eval_summary_lines:
                f.write(line)

    # 1.6 Plot coverage vs cutoff (Consensus) per dataset (optional, masih pakai full ep_cov_df)
    for ds_label in ep_cov_df["Dataset_label"].unique():
        sub = ep_cov_df[ep_cov_df["Dataset_label"] == ds_label].sort_values("Cutoff")
        plt.figure(figsize=(6, 4))
        plt.plot(sub["Cutoff"], sub["Consensus_IN_AD_pct"],
                 marker="o", linestyle="-", color="tab:blue",
                 label="Consensus IN_AD %")
        plt.axvline(tau_endpoint, color="red", linestyle="--",
                    label=f"Selected tau = {tau_endpoint:.2f}")
        plt.ylim(0, 100)
        plt.xlabel("Tanimoto cutoff (t)")
        plt.ylabel("Consensus AD coverage (%)")
        plt.title(f"{endpoint_name}\n{ds_label}")
        plt.grid(True, alpha=0.3)
        ax = plt.gca()
        ax.spines["top"].set_visible(False)
        ax.spines["right"].set_visible(False)
        plt.legend(frameon=False)
        plt.tight_layout()
        safe_ds = ds_label.replace(" ", "_").replace("–", "-").replace(":", "").replace("/", "_")
        plot_path = os.path.join(
            output_dir,
            f"{file_tag}_{safe_ds}_Consensus_AD_coverage_vs_cutoff_with_tau.png"
        )
        plt.savefig(plot_path, dpi=300)
        plt.close()
        print(f"  > Saved plot: {plot_path}")

print("\n✅ All endpoints AD processing finished (with data-driven tau_endpoint per endpoint, minimum cutoff 0.50, and evaluation-set summaries).")
```
